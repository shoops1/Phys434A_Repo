```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import scipy
from scipy import stats
from scipy.stats import norm
```

```python
# I will assume background cosmic radiation at X = 5 average for poisson distribution. I will assume the average gamma rays emitted from my source per day is Y = 50
```

```python
#Let's make a Poisson distribution for the background
background = stats.poisson.rvs(5,loc=0,size=10000)
```

```python
#Plot the poisson distribution we used as a histogram, and overlay the pmf for a poisson distribution at that expected average

plt.rcParams["figure.figsize"] = (15,10)
fig, ax = plt.subplots(1, 1)
ax.hist(background,40, density=True)
plt.tick_params(labelsize = 24)
plt.xlim([0.0,16])

p_line = np.linspace(0,16,1000)
ax.plot(p_line, stats.poisson.pmf(p_line,5,loc=0),linewidth = 8,alpha = 0.7)

plt.show()
```

```python
# How does the probability distribution change as we add more days observed? (show that 5 days follows poisson distribution still)
```

```python
# More days means more data, 5 days would be 5x as much data so we can represent as:

bg_5day = np.random.poisson(5,50000) #5x as many measurements
```

```python
# Now we will look at the question of averaging over days -> aka what are the odds we get this measurement 2,3,4,... days in a row
```

```python
# Use numpy convolve

x_range = np.arange(stats.poisson.ppf(0.01, 5),
              stats.poisson.ppf(0.99, 5))
pmf = stats.poisson.pmf(k=x_range,mu=5)
print(pmf)
conv = scipy.signal.convolve(pmf, pmf, mode='full')
plt.scatter(x_range,pmf)
```

```python
plt.rcParams["figure.figsize"] = (15,10)
fig, ax = plt.subplots(1, 1)
#ax.hist(conv,40, density=True)
#plt.tick_params(labelsize = 24)
#plt.xlim([0.0,25])

#p_line = np.linspace(0,16,1000)
ax.plot(x_range,pmf)

plt.show()
```
